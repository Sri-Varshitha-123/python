```python
#7th answer -

review = "The product quality is Excellent but delivery was Late"

# 1. Convert to lowercase
lower_review = review.lower()
print(lower_review)

# 2. Count number of words
word_count = len(review.split())
print("Number of words:", word_count)

# 3. Check if the word "excellent" exists
print("excellent" in lower_review)

# 4. Replace "Late" with "delayed"
new_review = review.replace("Late", "delayed")
print(new_review)
```

```python
#8th answer -
sales = [12000, 15000, 11000, 18000, 17000]

# 1. Calculate total revenue
total_revenue = sum(sales)
print("Total Revenue:", total_revenue)

# 2. Find average revenue
average_revenue = total_revenue / len(sales)
print("Average Revenue:", average_revenue)

# 3. Identify highest sales month
highest_sale = max(sales)
month = sales.index(highest_sale) + 1
print("Highest Sales Month:", month)

# 4. Append new month sale
sales.append(20000)
print("Updated Sales List:", sales)
# 5. Sort list in descending order
sales.sort(reverse=True)
print("Descending Order:", sales)
```

```python
#9th answer -
products = ["Laptop", "Phone", "Tablet"]
prices = [50000, 20000, 15000]
quantities = [5, 10, 7]

# 1. Compute total revenue for each product
revenues = []

for i in range(len(products)):
    revenue = prices[i] * quantities[i]
    revenues.append(revenue)

# 2. Identify product with maximum revenue
max_revenue = max(revenues)
max_product = products[revenues.index(max_revenue)]

# 3. Store revenues in a new list
print("Revenues List:", revenues)

# 4. Print formatted summary
print("\nSales Summary")
for i in range(len(products)):
    print(products[i], "-> Revenue =", revenues[i])

print("\nProduct with Maximum Revenue:", max_product)
print("Maximum Revenue:", max_revenue)
```
